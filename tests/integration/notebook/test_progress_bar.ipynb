# Test: progress bars are shown in a notebook

A kernel replaces `sys.stderr` with a stream that reports `isatty() == False`, which is what
`tqdm`'s `disable=None` uses to hide a bar. The kernel renders the redraws as an animation, so
the bar must be shown here anyway.

```python
import sys

from aiida.common.progress_reporter import get_progress_reporter, set_progress_bar_tqdm

assert not sys.stderr.isatty(), 'the kernel stderr is expected not to be a terminal'

set_progress_bar_tqdm()
with get_progress_reporter()(total=3, desc='Doing something') as progress:
    for _ in range(3):
        progress.update()
```
